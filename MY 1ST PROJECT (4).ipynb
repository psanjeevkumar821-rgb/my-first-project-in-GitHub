```python
import pandas as pd
```

```python
df=student_Details_data={
                "name":["mayaa","pranyaa","sanjeev","maha","saravana"],
                "python":[87,96,89,89,50],
                "pandas":[77,98,97,45,76],
                "ml":[90,91,98,36,78]

}
```

```python
df=pd.DataFrame(student_Details_data)
print(df.dtypes)
```

```python
print(df)
print(df.dtypes)
```

```python
df["total"]=df[["python","pandas","ml"]].sum(axis=1)
print(df["total"])
print(df.dtypes)
```

```python
df["average"]=df[["python","pandas","ml"]].mean(axis=1)
print(df["average"])
print(df.dtypes)
```

```python
df["result"]=df["average"].apply(lambda x:"pass" if x >=40 else "fail")
print(df["result"])
```

```python
print(df)
```

```python
print("\n---Student Performance Analysis---")
print(df)
```

```python
topper=df.loc[df["total"].idxmax()]
print("\n---class topper---")
print(topper["name"])
print("total mark:",topper["total"])
```

```python
print("\n---class average---")
print(round(df["average"].mean(),2))
```

```python
print("\n---pass student---")
print([df["result"]=="pass"])
```

```python
print("\n---rank list---")
print(df.sort_values(by="total"))
```

```python
df.to_csv("student_results.csv",index=False)
```
